```python
import requests
```

```python
url = "https://api.worldbank.org/v2/country/KE/indicator/DT.DOD.DECT.CD?format=json"

params = {
    'date': '2010:2024'
}
response = requests.get(url, params=params)
response.status_code
```

```python
data = response.json()
data
```

```python
import pandas as pd
```

```python
rel_data = []

for item in data[1]:
    rel_data.append({
        'country_code' : item['country']['id'],
        'debt_value' : item['value'],
        'year' : int(item['date']),
        'debt_desc' : item['indicator']['value']
    })


df = pd.DataFrame(rel_data)
df.dropna(inplace=True)
df
```
